```python
#Create a dataset with the following columns-> Products,Regions,Sales,Profit and 10 rows: 

import pandas as pd

# Create dataset
data = {
    'Products': ['Laptop', 'Mobile', 'Tablet', 'Laptop', 'Mobile',
                 'Tablet', 'Laptop', 'Mobile', 'Tablet', 'Laptop'],
    'Regions': ['North', 'South', 'East', 'West', 'North',
                'South', 'East', 'West', 'North', 'South'],
    'Sales': [50000, 30000, 20000, 55000, 32000,
              21000, 48000, 35000, 23000, 60000],
    'Profit': [8000, 5000, 3000, 9000, 5200,
               3200, 7500, 5800, 3400, 10000]
}

df = pd.DataFrame(data)

df
```

```python
# Plot a line chart of Sales for all 10 rows (index as X-axis). 

import seaborn as sns
import matplotlib.pyplot as plt 

#plt.figure()

plt.plot(df.index,df['Sales'])
plt.xlabel('Index')
plt.ylabel('Sales')
plt.title("line chart of Sales for all 10 rows")
plt.show()
```

```python
# Plot a bar chart showing total sales for each product. 
product_sales=df.groupby('Products')['Sales'].sum()
product_sales

plt.bar(product_sales.index , product_sales.values)
plt.xlabel('Product')
plt.ylabel('Sales')
plt.title("Total sales for each product")
```

```python
#Plot a scatter plot of Sales vs Profit. 
plt.scatter(x='Sales',y='Profit',data=df)
plt.title("Sales vs Profit")
```

```python
# Create a pie chart showing total sales by region. 

region_sales=df.groupby('Regions')['Sales'].sum()
region_sales
plt.bar(region_sales.index,region_sales.values)
```

```python
# Plot a histogram of the Sales column. 
plt.hist('Sales',data=df)
plt.xlabel('Sales')
plt.ylabel('Frequency')
plt.title('Histogram of Sales')
```

```python
#Create a boxplot of Profit to detect outliers. 
plt.boxplot('Profit',data=df)
plt.ylabel('Profit')
plt.title('Boxplot of Profit')
plt.show()
```

```python
# Create a countplot showing how many products are sold in each region. 
sns.countplot(x='Regions',hue='Products',data=df)
df
```

```python
#Create a heatmap to show correlation between Sales and Profit. 

corr = df[['Sales', 'Profit']].corr()
sns.heatmap(corr,annot=True,cmap='coolwarm')
```

```python
# Show total and average sales per region using pandas. 

sales_region=df.groupby('Regions')['Sales'].agg(['sum','mean'])
sales_region
```
