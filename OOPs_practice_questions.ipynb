```python
#Create a program that removes all duplicate characters from a string. 

def dup_char(text):
    new= ''
    for i in text:
        if i not in new:
            new += i
    return new

word=input("Enter string:")
print(dup_char(word))
```

```python
# Write a script to count the frequency of each character in a string. 

def freq_c(text):
    new={}
    for i in text:
        if i in new:
            new[i] += 1
        else:
            new[i]=1
    return new

word=input("Enter string:")
print(freq_c(word))
```

```python
#Write a program that accepts a sentence and calculates the number of upper and lower case letters. 

sentence = input("Enter a sentence: ")

upper_count = 0
lower_count = 0

for ch in sentence:
    if ch.isupper():
        upper_count += 1
    elif ch.islower():
        lower_count += 1

print("Uppercase letters:", upper_count)
print("Lowercase letters:", lower_count)
```

```python
# Create a program to find the longest word in a sentence. 

sentence = input("Enter a sentence: ")

word=sentence.split()
longest_word=word[0]
for i in word:
    if len(i)>len(longest_word):
        longest_word=i

print(longest_word)
```

```python
#Write a program that takes a string and returns the string in reverse order without using [::-1]. 
def reverse_s(word):
    new=''
    for i in word:
        new = i + new

    return new


word=input("Enter string:")
print(reverse_s(word))
```

```python
# Create a program that extracts numbers from a string and returns their sum.
def sum_of_numbers(text):
    total = 0
    current_number = ""

    for char in text:
        if char.isdigit():
            current_number += char
        else:
            if current_number != "":
                total += int(current_number)
                current_number = ""

    # add last number if string ends with a digit
    if current_number != "":
        total += int(current_number)

    return total


# Example
sentence = "I have 3 apples, 10 bananas and 5 oranges"
result = sum_of_numbers(sentence)

print("String:", sentence)
print("Sum of numbers:", result)

    
```

```python
# Create a script that takes a sentence and removes all stop words.

para="This is a simple sentence for learning data science in Python"
stop_words = {"is", "a", "the", "and", "in", "of", "to", "for", "this"}

new=[]
words=para.split()
for word in words:
    if word.lower() not in stop_words:
        new.append(word)

sentence= " ".join(new)
print(sentence)
```

```python
#Write a function to remove punctuation from a string. 
import string

def no_pun(text):
    new=''
    for i in text:
        if i not in string.punctuation:
            new += i

    return new

s = "Hello, world! Welcome to Python."
print(no_pun(s))
```

```python
#Write a function to generate acronyms from a sentence. 
def generate_acronym(sentence):
    words = sentence.split()
    acronym = ""

    for word in words:
        acronym += word[0].upper()

    return acronym

text = "data science and machine learning"
print(generate_acronym(text))

```

```python
#Write a script to check if a file contains a specific word. 

filename = "sample.txt"
search_word = input("Enter the word to search: ")

try:
    with open(filename, "r") as file:
        content = file.read()

        if search_word in content:
            print(f"'{search_word}' is present in the file.")
        else:
            print(f"'{search_word}' is NOT present in the file.")

except FileNotFoundError:
    print("File not found. Please check the file name.")
```

```python
#Write a program to calculate the average word length in a sentence. 

sentence=input("Enter one sentence:")
words=sentence.split()
total_char=sum(len(word) for word in words)

if len(words) >0:
    average_length=total_char/len(words)
    print(f"Average word length: {average_length}")
else:
    print("No words found in the sentence.")
    
```

```python
#Write a script that finds all email addresses in a given text. 
import re

text = """
Hello John, please contact me at john.doe@example.com. 
Also, you can reach Jane at jane_smith123@company.org or support@service.co.in.
"""

email_pattern=r'[a-zA-Z0-9._%+-]+@[a-zA-Z0-9._]+\.[[a-zA-Z]{2,}'
emails=re.findall(email_pattern,text)

if emails:
    print("Email address found")
    for email in emails:
        print(email)
else:
    print("No email addresses found.")
```

```python
#Create a function that validates an email address format. 
import re

def is_valid_email(email):
    email_pattern=r'[a-zA-Z0-9._%+-]+@[a-zA-Z0-9._]+\.[[a-zA-Z]{2,}'
    if re.match(email_pattern,email):
         return True
    else:
        return False

mails = ["john.doe@example.com", "invalid_email@com", "jane_smith@company.org"]
for e in mails:
    if is_valid_email(e):
        print(f"{e} is valid")
    else:
        print(f"{e} is not valid")
```
